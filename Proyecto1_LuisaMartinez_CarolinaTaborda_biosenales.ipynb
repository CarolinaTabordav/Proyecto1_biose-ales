#### **Historia de las BCI**

Las interfaces cerebro-computador (BCI) tienen sus raíces en el desarrollo de la electroencefalografía (EEG). A inicios del siglo XX, el neurólogo Hans Berger logró registrar por primera vez la actividad eléctrica del cerebro, identificando ondas como alfa y beta, lo que marcó el inicio del estudio sistemático de las señales cerebrales medibles. Este avance permitió comprender que la actividad neuronal podía ser capturada, analizada y, potencialmente, utilizada como medio de comunicación.

A partir de este descubrimiento, durante las décadas de 1950 y 1960 se evidenció que los seres humanos podían aprender a modular voluntariamente sus propias ondas cerebrales mediante técnicas de neurofeedback. Este hallazgo fue clave, ya que demostró que la actividad cerebral no solo podía registrarse, sino también controlarse de forma consciente, sentando así las bases para el desarrollo de sistemas interactivos basados en señales neuronales.

En este contexto, en 1973 el investigador Jacques Vidal introdujo formalmente el concepto de interfaz cerebro-computador, proponiendo el uso directo de señales cerebrales para controlar dispositivos externos, como computadoras. Esta propuesta representó un cambio de paradigma, al plantear una vía de comunicación directa entre el cerebro y las máquinas, sin necesidad de la intervención del sistema muscular.

Posteriormente, en 1988 se desarrolló uno de los primeros sistemas funcionales de BCI, conocido como P300 Speller, el cual permitía a los usuarios seleccionar letras mediante la detección de potenciales evocados en el EEG. Este sistema demostró la viabilidad práctica de las BCI para la comunicación, especialmente en personas con limitaciones motoras severas. A partir de entonces, el campo ha experimentado un crecimiento significativo, impulsado por avances en el procesamiento de señales, la computación y, más recientemente, la inteligencia artificial [1].



#### **Usos actuales de las BCI**

Las interfaces cerebro-computador (BCI) tienen actualmente múltiples aplicaciones en distintos campos, destacándose especialmente en la medicina y la rehabilitación. En este contexto, se emplean para restaurar funciones motoras y comunicativas en pacientes con afectaciones neurológicas, permitiendo, por ejemplo, la comunicación en personas con parálisis o enfermedades neurodegenerativas, el control de prótesis y dispositivos robóticos, y la rehabilitación motora tras accidentes cerebrovasculares. Asimismo, se han desarrollado terapias basadas en neurofeedback, en las que los usuarios aprenden a modular su actividad cerebral con fines terapéuticos.

Adicionalmente, las BCI han ampliado las posibilidades de interacción humano-máquina al permitir una comunicación directa entre el cerebro y dispositivos externos sin necesidad de actividad muscular. Esto ha dado lugar a aplicaciones como el control de computadoras mediante señales cerebrales, el desarrollo de entornos interactivos y sistemas de asistencia tecnológica orientados a mejorar la accesibilidad y la autonomía de los usuarios.

Por último, estas tecnologías también se utilizan en el monitoreo de estados cognitivos, facilitando la evaluación de variables como la atención, la fatiga y el estrés. Este tipo de aplicaciones tiene un impacto creciente en ámbitos como la educación, el entorno laboral y el deporte, donde el análisis de la actividad cerebral puede contribuir a optimizar el rendimiento y el bienestar.

#### **Proyectos futuros**

Las interfaces cerebro-computador (BCI) han experimentado un crecimiento en los últimos años, orientándose hacia el desarrollo de tecnologías cada vez más precisas, accesibles y adaptables. Uno de los principales enfoques a futuro es el perfeccionamiento de las BCI no invasivas basadas en electroencefalografía (EEG). Aunque estas presentan ventajas en términos de seguridad y facilidad de uso, aún enfrentan limitaciones como la baja resolución espacial y la susceptibilidad al ruido. Por ello, la investigación actual se centra en mejorar la calidad de adquisición de señales, el diseño de electrodos más eficientes y el desarrollo de algoritmos robustos que permitan una interpretación más confiable de la actividad cerebral [1][2].




En este contexto, la integración con técnicas de inteligencia artificial, particularmente machine learning y deep learning, se perfila como un elemento clave para el avance del campo. Estas herramientas permiten mejorar la clasificación de patrones neuronales, reducir los tiempos de entrenamiento de los usuarios y desarrollar sistemas más adaptativos y generalizables entre individuos. Investigaciones recientes dentro del campo de la neuroingeniería destacan que el uso de modelos de aprendizaje profundo ha incrementado significativamente la precisión en tareas como la detección de imaginería motora, lo que abre la puerta a aplicaciones más eficientes en rehabilitación y control de dispositivos [3].


A mediano y largo plazo, se espera que las BCI trasciendan el ámbito clínico y se integren en la vida cotidiana. El desarrollo de dispositivos portátiles o wearables permitirá su uso en áreas como la educación, el entretenimiento, el monitoreo cognitivo y la interacción humano-máquina [4]. Empresas como Neuralink y Synchron están impulsando avances tanto en interfaces invasivas como no invasivas, buscando no solo restaurar funciones perdidas en pacientes, sino también ampliar las capacidades humanas mediante la comunicación directa con sistemas computacionales [5][6].

Otra tendencia relevante es el desarrollo de interfaces cerebro-computador bidireccionales, capaces no solo de interpretar señales cerebrales, sino también de estimular el sistema nervioso. Este enfoque tiene aplicaciones prometedoras en el tratamiento de enfermedades neurológicas como la epilepsia, el parkinson o la esclerosis lateral amiotrófica, así como en la restauración sensorial[7].




Sin embargo, el avance de las BCI también plantea importantes desafíos éticos y sociales. La posibilidad de acceder, interpretar e incluso modificar la actividad cerebral genera preocupaciones relacionadas con la privacidad de los datos neuronales, la seguridad de la información y el consentimiento informado. Organizaciones como el IEEE han comenzado a establecer marcos éticos para el desarrollo responsable de la neurotecnología, destacando la necesidad de regular su uso antes de una adopción masiva [8].


#### **Actividad cerebral registrada por EEG durante movimientos e imaginación de movimiento de las manos**

La actividad cerebral asociada al movimiento y a la imaginería motora se estudia ampliamente mediante electroencefalografía (EEG), una técnica no invasiva que registra la actividad eléctrica cerebral a través de electrodos colocados sobre el cuero cabelludo. Estos electrodos se ubican siguiendo el sistema internacional 10–20 o 10–10, que estandariza su posición en relación con áreas corticales específicas. En el caso del control motor de las manos, los electrodos más relevantes son C3, Cz y C4, localizados sobre la corteza motora primaria. Además, en sistemas BCI es común emplear configuraciones multicanal (por ejemplo, 8, 16 o más electrodos), así como montajes referenciales o bipolares, dependiendo del tipo de análisis. También se distinguen electrodos húmedos, que utilizan gel conductor y ofrecen mejor calidad de señal, y electrodos secos, que facilitan la portabilidad aunque con mayor susceptibilidad al ruido [9].

Durante la ejecución o imaginación de movimientos de las manos, se observan cambios en los ritmos sensorimotores, principalmente en las bandas mu (8–13 Hz) y beta (13–30 Hz). Estos ritmos presentan una disminución de potencia durante la actividad motora, fenómeno conocido como event-related desynchronization (ERD), seguido en algunos casos por un aumento posterior denominado event-related synchronization (ERS). Estos patrones reflejan la activación y desactivación de poblaciones neuronales involucradas en la planificación y ejecución del movimiento. En tareas de imaginería motora, estos cambios ocurren incluso sin movimiento físico, lo que permite su uso en sistemas BCI [10][11].

En tareas específicas como la imaginería motora de manos derecha e izquierda, se observa un patrón espacial característico, donde la actividad EEG presenta desincronización del ritmo mu en el hemisferio contralateral al movimiento (por ejemplo, imaginar mover la mano derecha afecta principalmente el hemisferio izquierdo). Este patrón es fundamental en sistemas BCI, ya que permite distinguir entre diferentes intenciones motoras. Estudios han demostrado que tanto el movimiento real como el imaginado generan patrones similares en estos ritmos, lo que hace posible utilizar la imaginería motora como base para controlar dispositivos externos [12].

El análisis de señales EEG en sistemas BCI realiza un preprocesamiento que incluye filtrado en las bandas de interés (8–30 Hz) y eliminación de artefactos. Luego, se lleva a cabo la extracción de características, donde técnicas como Common Spatial Patterns (CSP) permiten resaltar diferencias espaciales entre clases (por ejemplo, mano derecha vs izquierda). Posteriormente, estas características son ingresadas a algoritmos de clasificación como análisis discriminante lineal (LDA) o máquinas de soporte vectorial (SVM), que permiten identificar la intención del usuario en tiempo real. Finalmente, la salida del clasificador se traduce en comandos para controlar dispositivos externos, como cursores, prótesis o interfaces gráficas.


#### **Flujos de procesamiento**

El desarrollo de sistemas de interfaz cerebro-computador (BCI) basados en señales EEG sigue un flujo de procesamiento estructurado que permite transformar señales biológicas complejas en comandos interpretables por un sistema. Este flujo, ampliamente documentado en la literatura, consta de varias etapas: adquisición de señales, preprocesamiento, extracción de características, clasificación y generación de salida [4] [9].

La primera etapa corresponde a la adquisición de señales EEG mediante sistemas multicanal, usualmente configurados bajo el sistema internacional 10–20 [13]. En el contexto de un curso como Bioseñales y Simulación de Sistemas, esta etapa puede abordarse tanto desde datos como mediante señales simuladas que reproduzcan características espectrales de los ritmos cerebrales. La simulación permite comprender fenómenos como ruido, artefactos y variabilidad intersujeto sin depender exclusivamente de hardware especializado[14].

Posteriormente, el preprocesamiento tiene como objetivo mejorar la calidad de la señal. Esto incluye filtrado en bandas de interés (por ejemplo, 8–30 Hz para ritmos mu y beta), eliminación de artefactos (parpadeo, movimiento muscular) y segmentación temporal de las señales. Para los casos como el proyecto de este curso, estas operaciones pueden implementarse mediante herramientas como Python, utilizando procesamiento digital de señales y operaciones matemáticas a través de librerías como numpy, scipy o mn [10].

La etapa de extracción de características es crítica, ya que determina la capacidad del sistema para distinguir entre diferentes estados mentales. En sistemas basados en imaginería motora, técnicas como Common Spatial Patterns (CSP) permiten maximizar la discriminación entre clases (por ejemplo, mano derecha vs izquierda), mientras que métodos basados en potencia espectral permiten analizar la energía en bandas específicas. Estas técnicas son especialmente relevantes para la simulación del sistema, ya que permiten comparar resultados con diferentes configuraciones [15].

En la etapa de clasificación, las características extraídas a partir de las señales EEG se utilizan para asignar cada segmento de señal a una clase específica, como imaginería de mano derecha, izquierda o estado de reposo. Para ello, se emplean algoritmos sencillos para los análisis lineales y vectoriales que se requieran que requieran pocos recursos computacionales y demuestren buen desempeño en problemas de imaginería motora con EEG, donde la relación señal-ruido sea baja. Finalmente, la salida del sistema consiste en la etiqueta predicha para cada segmento de señal, la cual puede representarse de forma simple mediante gráficos o matrices de confusión que permitan evaluar el desempeño del modelo. En este tipo de implementación, la salida no se traduce en el control de dispositivos físicos, sino en la capacidad de diferenciar correctamente entre las clases analizadas, lo que permite validar el funcionamiento del sistema BCI a partir de métricas como la exactitud o el error de clasificación [16].

```python
import os
import re
from pathlib import Path

# 1. Configuración dinámica
CANTIDAD_SUJETOS = 50
sujetos_objetivo = [f"{i:03d}" for i in range(1, CANTIDAD_SUJETOS + 1)]
archivos_encontrados = []

MAPEO_RUNS = {
    **{r: 'Real' for r in [3, 4, 7, 8, 11, 12]},
    **{r: 'Imagery' for r in [5, 6, 9, 10, 13, 14]}
}

print(f"--- INICIANDO BÚSQUEDA DE ARCHIVOS (Sujetos 001 al {CANTIDAD_SUJETOS:03d}) ---")

# 2. Búsqueda optimizada
base_path = Path('.') # Directorio actual

for archivo in base_path.rglob('*.set'):
    nombre = archivo.name
    

    if nombre.startswith('._'):
        continue
    
    nombre_low = nombre.lower()
    
    # Extraer ID de sujeto y Run con Regex
    sub_match = re.search(r'sub-(\d+)', nombre_low)
    run_match = re.search(r'run-(\d+)', nombre_low)
    
    if sub_match and run_match:
        id_sub = sub_match.group(1)
        run_num = int(run_match.group(1))
        
  
        if id_sub in sujetos_objetivo and run_num in MAPEO_RUNS:
            archivos_encontrados.append({
                'subject': f"sub-{id_sub}",
                'run': run_num,
                'task_type': MAPEO_RUNS[run_num],
                'path': str(archivo)
            })

# 3. Reporte de resultados
total = len(archivos_encontrados)
print(f"¡Éxito! Se encontraron {total} archivos válidos.")

if total > 0:
  
    from collections import Counter
    conteo = Counter([a['task_type'] for a in archivos_encontrados])
    print(f"Distribución: {conteo['Real']} Reales | {conteo['Imagery']} Imaginarios")
    
    print("\nEjemplo del último sujeto encontrado:")
    print(f"Sujeto: {archivos_encontrados[-1]['subject']} | Tarea: {archivos_encontrados[-1]['task_type']}")
else:
    print("\n[!] ADVERTENCIA: No se encontraron archivos. Revisa la estructura de carpetas.")
```

### 1. Función de procesamiento 

```python
import numpy as np
from scipy.signal import firwin, filtfilt, welch

FS = 160.0 
NPERSEG = int(FS * 2) 

# Diccionario de bandas de frecuencia
BANDAS = {
    'delta': (1, 4),
    'theta': (4, 8),
    'mu': (8, 13),    
    'beta': (13, 30),
    'gamma': (30, 45)
}
import numpy as np
from scipy.signal import firwin, filtfilt, welch

import numpy as np
from scipy.signal import firwin, filtfilt, welch

def procesar_eeg(signal, fs, verbose=False):
    """
    Versión unificada para Bioingeniería (Compatible con Python 3.14 y Matplotlib)
    """
    # 1. Señal cruda (la guardamos tal cual)
    raw_signal = signal

    # 2. Filtro FIR pasabanda 1-45 Hz
    numtaps = 101
    b = firwin(numtaps, [1.0, 45.0], pass_zero=False, fs=fs, window='hamming')
    filtered = filtfilt(b, [1.0], signal)


    detrended = filtered - np.mean(filtered)

    # 4. Cálculo de PSD (Welch)
    freqs, psd = welch(detrended, fs=fs, nperseg=int(2*fs), window='hann')

    # 5. Potencias por banda (con corrección para Python 3.14)
    def integrar(y, x):
        if hasattr(np, 'trapezoid'): return np.trapezoid(y, x)
        return np.trapz(y, x)

    potencias = {}
   
    for banda, (flo, fhi) in BANDAS.items():
        mask = (freqs >= flo) & (freqs <= fhi)
        potencias[banda] = integrar(psd[mask], freqs[mask])

    if verbose:
        print("\nPotencias calculadas:")
        for k, v in potencias.items(): print(f"  {k}: {v:.4e} µV²")

    return {
        'raw': raw_signal,
        'filtered': filtered,
        'detrended': detrended,
        'freqs': freqs,
        'psd': psd,
        'potencias': potencias
    }

print("Función procesar_eeg sincronizada con el código de graficación.")
```

```python
import mne
import matplotlib.pyplot as plt
import numpy as np

CANALES_MOTOR = ['C3', 'Cz', 'C4']


meta_demo = next((a for a in archivos_encontrados if a['task_type'] == 'Imagery'), archivos_encontrados[0])

print(f"--- VISUALIZACIÓN MULTICANAL ---")
print(f"Archivo: {meta_demo['subject']} | Run: {meta_demo['run']} | Tarea: {meta_demo['task_type']}")


raw = mne.io.read_raw_eeglab(meta_demo['path'], preload=True, verbose=False)
raw.rename_channels({ch: ch.rstrip('.') for ch in raw.ch_names})

# 4. Extraer el primer evento T1 (inicio de tarea)
eventos, mapa = mne.events_from_annotations(raw, verbose=False)
id_interes = next((codigo for nombre, codigo in mapa.items() if 'T1' in nombre or 'left' in nombre.lower()), None)

if id_interes is None:
    id_interes = eventos[0][2]

primer_evento = next(ev for ev in eventos if ev[2] == id_interes)
onset_muestras = primer_evento[0]
FS_REAL = raw.info['sfreq']
muestras_duracion = int(4.0 * FS_REAL)

# 5. Bucle para generar una gráfica por cada canal
for ch_name in CANALES_MOTOR:
    if ch_name not in raw.ch_names:
        print(f"Saltando {ch_name}: No encontrado en este archivo.")
        continue
    
    # Extraer segmento del canal actual
    idx_canal = raw.ch_names.index(ch_name)
    data_v, _ = raw[idx_canal, onset_muestras : onset_muestras + muestras_duracion]
    señal_uV = data_v[0] * 1e6
    t_demo = np.arange(len(señal_uV)) / FS_REAL


    res = procesar_eeg(señal_uV, fs=FS_REAL, verbose=False)

    fig, axes = plt.subplots(4, 1, figsize=(12, 10), constrained_layout=True)
    fig.suptitle(f'ANÁLISIS DE CORTEZA MOTORA\nSujeto: {meta_demo["subject"]} | Canal: {ch_name} | Tarea: {meta_demo["task_type"]}', 
                 fontsize=14, fontweight='bold')

    # Panel 1: Cruda
    axes[0].plot(t_demo, res['raw'], color='steelblue', lw=0.7)
    axes[0].set_title(f'1. Señal Cruda - Canal {ch_name}')
    axes[0].set_ylabel('µV')

    # Panel 2: Filtrada
    axes[1].plot(t_demo, res['filtered'], color='darkorange', lw=0.7)
    axes[1].set_title('2. Filtrada Pasabanda (1–45 Hz)')
    axes[1].set_ylabel('µV')

    # Panel 3: Sin Media
    axes[2].plot(t_demo, res['detrended'], color='seagreen', lw=0.7)
    axes[2].set_title('3. Señal con Remoción de Media (Zero-mean)')
    axes[2].set_ylabel('µV')

    # Panel 4: PSD (Potencia)
    ax4 = axes[3]
    ax4.semilogy(res['freqs'], res['psd'], color='purple', lw=1.5)
    colores_banda = {'delta': '#AED6F1', 'theta': '#A9DFBF', 'mu': '#F9E79F', 'beta': '#F1948A', 'gamma': '#D7BDE2'}

    for banda, (flo, fhi) in BANDAS.items():
        ax4.axvspan(flo, fhi, color=colores_banda[banda], alpha=0.3, label=banda)

    ax4.set_xlim(0, 50)
    ax4.set_title(f'4. Densidad Espectral de Potencia - Canal {ch_name}')
    ax4.set_xlabel('Frecuencia (Hz)')
    ax4.set_ylabel('µV²/Hz')
    ax4.legend(loc='upper right', ncol=5, fontsize='x-small')

    plt.show()
```

En la señal cruda se observa un comportamiento bastante irregular, con presencia de picos de gran amplitud (alrededor de ±80–100 µV), lo cual es típico en registros EEG pero también puede indicar la presencia de artefactos como movimientos, parpadeos o actividad muscular. Además, la señal no está completamente centrada en cero, lo que sugiere un pequeño desplazamiento de la línea base. 
Tras aplicar el filtrado pasabanda entre 1 y 45 Hz, la señal se vuelve más estable y limpia, eliminando componentes de muy baja y alta frecuencia, aunque aún se mantienen algunos picos, lo que indica que ciertos artefactos no han sido completamente removidos. Posteriormente, al aplicar la remoción de la media, la señal queda centrada en cero sin alterar su forma, lo cual es un paso importante para evitar sesgos en análisis posteriores, especialmente en el dominio frecuencial. 
Finalmente, en la densidad espectral de potencia se observa que la energía de la señal está distribuida en varias bandas de frecuencia, sin un pico dominante claro en la banda alfa (8–13 Hz), mientras que se aprecia una mayor presencia en beta (13–30 Hz) y una disminución progresiva hacia frecuencias más altas. Esto sugiere que la señal corresponde a un estado activo o con influencia de ruido, más que a un estado de reposo claramente definido.

### 2. Densidad espectral de potencia 

```python
import mne
import numpy as np
from scipy.signal import welch
import pandas as pd


CANALES_INT = ['C3', 'Cz', 'C4']
LABEL_MAP = {
    'T1': 'Izquierda', 
    'T2': 'Derecha',
    'T0': 'Reposo'
}

def calcular_psd_por_bandas(signal, fs):
    """
    Calcula la PSD y extrae la potencia media en bandas específicas.
    """
    BANDAS = {
        'mu': (8, 13),
        'beta': (13, 30)
    }
    
    # nperseg: tamaño de ventana (2 segundos para buena resolución en 160Hz)
    nperseg = min(len(signal), int(2 * fs))
    
    # Calculamos la densidad espectral de potencia
    freqs, psd = welch(signal, fs=fs, nperseg=nperseg, window='hann')
    
    resultados_bandas = {}
    for nombre, (f_min, f_max) in BANDAS.items():
        # Máscara booleana para seleccionar el rango de frecuencia
        idx_banda = np.logical_and(freqs >= f_min, freqs <= f_max)
        
        # Calculamos la potencia media (o integral) en esa banda específica
        if np.any(idx_banda):
            potencia_media = np.mean(psd[idx_banda])
        else:
            potencia_media = 0.0
            
        resultados_bandas[f'psd_{nombre}'] = potencia_media
        
    return resultados_bandas

def leer_set_y_procesar_completo(filepath: str,
                                canales: list = CANALES_INT,
                                duracion_seg: float = 4.0) -> list:
    """
    Lee .set, segmenta por eventos y calcula PSD por bandas para cada segmento.
    """
    # Cargar archivo de datos reales (ignorando archivos ._ de 4KB)
    try:
        raw = mne.io.read_raw_eeglab(filepath, preload=True, verbose=False)
    except Exception as e:
        print(f"No se pudo leer {filepath}: {e}")
        return []

    # Limpieza de nombres de canales
    raw.rename_channels({ch: ch.rstrip('.') for ch in raw.ch_names})
    
    fs = int(raw.info['sfreq'])
    # Seleccionamos solo los canales que existen en el archivo y nos interesan
    ch_disponibles = [ch for ch in canales if ch in raw.ch_names]
    if not ch_disponibles:
        return []
    
    raw.pick(ch_disponibles)

    # Extraer eventos desde las anotaciones de EEGLAB
    try:
        eventos, mapa_eventos = mne.events_from_annotations(raw, verbose=False)
        mapa_id = {v: k for k, v in mapa_eventos.items()}
    except Exception:
        return []

    muestras_ventana = int(duracion_seg * fs)
    segmentos_procesados = []

    for ev in eventos:
        onset_muestras = ev[0]
        id_evento = ev[2]
 
        etiq_original = mapa_id.get(id_evento, '')
        etiqueta = LABEL_MAP.get(etiq_original, None)
        
        if etiqueta is None: 
            continue

        inicio = onset_muestras
        fin = inicio + muestras_ventana
        
        # Validar que el segmento no se salga del final del archivo
        if fin > raw.n_times: 
            continue

        data_epoch, _ = raw[:, inicio:fin]
        data_uv = data_epoch * 1e6 

        for i, ch in enumerate(ch_disponibles):
        
            caracteristicas_psd = calcular_psd_por_bandas(data_uv[i], fs)
            
        
            info_segmento = {
                'label': etiqueta,
                'canal': ch,
                'onset_sec': onset_muestras / fs,
                **caracteristicas_psd  # Agrega psd_mu y psd_beta al diccionario
            }
            segmentos_procesados.append(info_segmento)

    return segmentos_procesados

print("Pipeline de procesamiento de PSD finalizado y listo para usar.")
```

Se observa que, en general, ambas señales mantienen una distribución de energía similar en las frecuencias bajas y medias (aproximadamente entre 1 y 40 Hz), lo que indica que el procesamiento no altera significativamente la información relevante del EEG en estas bandas. Sin embargo, a partir de los 40–45 Hz se evidencia una diferencia importante, la señal procesada presenta una caída abrupta de la potencia, mientras que la señal cruda mantiene valores más elevados. Esto confirma que el filtrado aplicado está funcionando correctamente, eliminando componentes de alta frecuencia asociadas principalmente a ruido, como actividad muscular (EMG) o interferencias eléctricas. En las bajas frecuencias también se aprecia una ligera reducción de potencia en la señal procesada, lo que sugiere la eliminación de componentes de muy baja frecuencia.

### 3. Almacenamiento de datos 

```python
import pandas as pd
import numpy as np
import mne

def leer_y_segmentar_maestro(path_archivo, canales_interes=['C3', 'Cz', 'C4']):
    """
    Extrae segmentos de 4s y calcula potencias para tareas Reales e Imaginarias.
    """
    try:
        # 1. Cargar archivo y limpiar nombres de canales
        raw = mne.io.read_raw_eeglab(path_archivo, preload=True, verbose=False)
        raw.rename_channels({ch: ch.rstrip('.') for ch in raw.ch_names})
        fs = raw.info['sfreq']
        

        eventos, mapa = mne.events_from_annotations(raw, verbose=False)
        

        ids_validos = {}
        for nombre, codigo in mapa.items():
            if 'T1' in nombre: ids_validos[codigo] = 'Izquierda'
            if 'T2' in nombre: ids_validos[codigo] = 'Derecha'
            if 'T0' in nombre: ids_validos[codigo] = 'Reposo'

        segmentos_resultados = []
        duracion_muestras = int(4.0 * fs) 

        for ev in eventos:
            onset, _, codigo = ev
            if codigo in ids_validos:
                condicion = ids_validos[codigo]
                
                for ch in canales_interes:
                    if ch in raw.ch_names:
                        idx = raw.ch_names.index(ch)

                        data, _ = raw[idx, onset : onset + duracion_muestras]
                        señal = data[0] * 1e6 # Convertir a microvoltios
                        
            
                        res = procesar_eeg(señal, fs=fs, verbose=False)
                        
                        segmentos_resultados.append({
                            'Canal': ch,
                            'Condicion': condicion,
                            'Delta': res['potencias']['delta'],
                            'Theta': res['potencias']['theta'],
                            'Mu':    res['potencias']['mu'],
                            'Beta':  res['potencias']['beta'],
                            'Gamma': res['potencias']['gamma']
                        })
        return segmentos_resultados
    except Exception as e:
        return []

# --- PROCESAMIENTO DE LOS 240 ARCHIVOS ---
registros_finales = []
print(f"Procesando base de datos completa ({len(archivos_encontrados)} archivos)...")

for i, meta in enumerate(archivos_encontrados):
    # Procesamos el archivo (sea 'Motion' o 'Imagery')
    segmentos = leer_y_segmentar_maestro(meta['path'])
    
    for seg in segmentos:
        registros_finales.append({
            'Sujeto': meta['subject'],
            'Tarea':  meta['task_type'], 
            **seg # Agrega Canal, Condicion y Potencias
        })
    
    if (i + 1) % 20 == 0:
        print(f" Progreso: {i+1}/240 archivos analizados.")


df_eeg = pd.DataFrame(registros_finales)

if not df_eeg.empty:
    print(f"\n¡Listo! Se generó un dataset de {len(df_eeg)} filas.")
    # Ordenamos para que sea fácil de leer
    df_eeg = df_eeg.sort_values(by=['Sujeto', 'Tarea', 'Condicion'])
    
    # Mostramos una mezcla de filas para ver Real e Imaginario
    display(df_eeg.sample(20))
else:
    print("El DataFrame está vacío. Verifica que 'archivos_encontrados' tenga datos.")
```

### **a. Estadística descriptiva y comparación entre condiciones**

Se analizan las cuatro condiciones: **reposo (T0)**, **mano izquierda (T1)**, **mano derecha (T2)**, separadas por tipo de tarea (**Movimiento real** e **Imaginería motora**), en los electrodos C3, Cz y C4.



```python
import pandas as pd
import numpy as np


print("Columnas detectadas en df_eeg:", df_eeg.columns.tolist())

# Limpiamos nombres (quitar espacios y estandarizar a lo que pide el Punto 3)
df_eeg.columns = [str(c).strip() for c in df_eeg.columns]

# Mapeo de seguridad: buscamos nombres parecidos y los unificamos
mapping_seguro = {
    'Sujeto': 'Sujeto', 'subject': 'Sujeto',
    'Tarea': 'Tarea', 'task_type': 'Tarea',
    'Condicion': 'Condicion', 'condition': 'Condicion',
    'Canal': 'Canal', 'canal': 'Canal',
    'run': 'run', 'Run': 'run'
}

df_eeg = df_eeg.rename(columns=mapping_seguro)

# 2. DEFINICIÓN DE ETIQUETAS
COND_LABELS = {
    'Reposo': 'Reposo (T0)', 
    'Izquierda': 'Mano izq. (T1)', 
    'Derecha': 'Mano der. (T2)'
}


cols_finales = [c for c in ['Sujeto', 'run', 'Tarea', 'Condicion', 'Mu', 'Beta'] if c in df_eeg.columns]

print(f"Columnas que se usarán para el análisis: {cols_finales}")

df_c3 = df_eeg[df_eeg['Canal'] == 'C3'][cols_finales].copy()
df_c4 = df_eeg[df_eeg['Canal'] == 'C4'][cols_finales].copy()


on_cols = [c for c in ['Sujeto', 'run', 'Tarea', 'Condicion'] if c in df_c3.columns]

df_li = df_c3.merge(df_c4, on=on_cols, suffixes=('_c3', '_c4'))

# 5. CÁLCULO DE LATERALIDAD
df_li['LI_mu']   = (df_li['Mu_c3']   - df_li['Mu_c4'])   / (df_li['Mu_c3']   + df_li['Mu_c4'])
df_li['LI_beta'] = (df_li['Beta_c3'] - df_li['Beta_c4']) / (df_li['Beta_c3'] + df_li['Beta_c4'])


df_li['Condición'] = df_li['Condicion'].map(COND_LABELS)

print("\n--- PROCESO COMPLETADO ---")
print(f"Filas en df_li: {len(df_li)}")
if not df_li.empty:
    display(df_li.head())
```

Los valores de LI (tanto positivos como negativos) reflejan la dominancia relativa entre hemisferios: valores positivos indican mayor actividad en C3 (hemisferio izquierdo) y negativos en C4 (hemisferio derecho). Sin embargo, al ver los primeros registros, los valores de LI son cercanos a cero o con variaciones pequeñas, lo que sugiere que la lateralización no es muy fuerte ni consistente en estos datos.

Sin embargo, estas diferencias no son muy marcadas y existe una alta variabilidad (cajas amplias y bigotes largos), lo que indica que la señal presenta dispersión considerable entre segmentos.
En cuanto a la lateralización, que sería esperada en tareas motoras no se aprecia un patrón claramente definido en estos diagramas, ya que las distribuciones entre T1 y T2 son bastante similares dentro de cada canal. En la banda beta ocurre algo similar, aunque hay ligeras variaciones entre condiciones, no se observa una separación evidente entre reposo e imaginación ni entre tipos de movimiento. ### **b. Pruebas estadísticas**

**Hipótesis planteadas:**

**H1 — Banda mu en imaginería motora: mano derecha vs mano izquierda**
- H₀: No existen diferencias en la potencia mu entre imaginería derecha e izquierda en C3 y C4
- H₁: Existen diferencias en la potencia mu entre imaginería derecha e izquierda en C3 y C4

**H2 — Banda beta en imaginería motora: mano derecha vs mano izquierda**
- H₀: No existen diferencias en la potencia beta entre imaginería derecha e izquierda
- H₁: Existen diferencias en la potencia beta entre imaginería derecha e izquierda

**H3 — Efecto de condición (Kruskal-Wallis): rest vs left vs right**
- H₀: No existen diferencias entre las tres condiciones en mu y beta
- H₁: Al menos una condición difiere de las demás

```python
from scipy import stats

print("=" * 75)
print("PRUEBA DE NORMALIDAD — Shapiro-Wilk (muestra = min(50, n) por grupo)")
print("α = 0.05  →  p < 0.05 indica NO normalidad")
print("=" * 75)


subgrupos = [
    ('Motion',  'Reposo',    'C3', 'Mu'), 
    ('Motion',  'Derecha',   'C3', 'Mu'),
    ('Motion',  'Izquierda', 'C3', 'Mu'), 
    ('Imagery', 'Reposo',    'C3', 'Mu'),
    ('Imagery', 'Derecha',   'C3', 'Mu'), 
    ('Imagery', 'Izquierda', 'C3', 'Mu'),
    ('Motion',  'Derecha',   'C3', 'Beta'), 
    ('Imagery', 'Derecha',   'C3', 'Beta'),
    ('Motion',  'Izquierda', 'C4', 'Mu'), 
    ('Imagery', 'Izquierda', 'C4', 'Mu')
]

print(f"{'Tipo':<10} {'Cond':<10} {'Canal':<6} {'Banda':<6} "
      f"{'Stat':>10} {'p-valor':>10} {'Normal?':>10}")
print("-" * 75)

for tipo, cond, ch, banda in subgrupos:

    datos = df_eeg[(df_eeg['Tarea'] == tipo) & 
                   (df_eeg['Condicion'] == cond) & 
                   (df_eeg['Canal'] == ch)][banda].dropna()
    
    n = min(len(datos), 50)
    
    if n < 8:
        print(f"{tipo:<10} {cond:<10} {ch:<6} {banda:<6}  (datos insuficientes: n={len(datos)})")
        continue
    
    # Ejecutamos Shapiro-Wilk sobre la muestra
    stat, p = stats.shapiro(datos.iloc[:n].values)
    normal  = "✓ Sí" if p > 0.05 else "✗ No"
    
    print(f"{tipo:<10} {cond:<10} {ch:<6} {banda:<6} {stat:>10.4f} {p:>10.4f} {normal:>10}")

print("\n" + "=" * 75)
print("→ Conclusión: Si la mayoría de los grupos marcan '✗ No',")
print("  los datos NO siguen una distribución normal.")
print("  Se deben aplicar pruebas NO PARAMÉTRICAS (Mann-Whitney U, Kruskal-Wallis).")
print("=" * 75)
```

```python
import pandas as pd
from scipy import stats

# --- H1 y H2: Mann-Whitney U — imaginería mano derecha vs izquierda ---
print("=" * 75)
print("H1 / H2 — IMAGINERÍA MOTORA: MANO DERECHA vs MANO IZQUIERDA")
print("Prueba: Mann-Whitney U (bilateral)  |  α = 0.05")
print("=" * 75)

# Filtramos solo por Imaginería
img = df_eeg[df_eeg['Tarea'] == 'Imagery']
der = img[img['Condicion'] == 'Derecha']
izq = img[img['Condicion'] == 'Izquierda']

filas = []
for ch in ['C3', 'Cz', 'C4']:
    for banda in ['Mu', 'Beta']:
        # Filtramos datos por canal y banda
        g1 = der[der['Canal'] == ch][banda].dropna().values
        g2 = izq[izq['Canal'] == ch][banda].dropna().values
        
        if len(g1) < 5 or len(g2) < 5:
            continue
            

        stat, p = stats.mannwhitneyu(g1, g2, alternative='two-sided')
        decision = "✓ Rechaza H0" if p < 0.05 else "— No rechaza H0"
        
        filas.append({
            'Canal': ch, 
            'Banda': banda, 
            'n_Derecha': len(g1),
            'n_Izquierda': len(g2), 
            'U_stat': round(stat, 1),
            'p-valor': round(p, 4), 
            'Decisión (α=0.05)': decision
        })


display(pd.DataFrame(filas))
```

El resultado de la prueba Mann-Whitney U muestra que no existen diferencias estadísticamente significativas entre la imaginación de la mano derecha e izquierda en ninguno de los canales (C3, Cz, C4) ni en las bandas mu y beta, ya que todos los valores p son mayores a 0.05. Esto implica que no se rechaza la hipótesis nula (H0) en todos los casos, es decir, las distribuciones de potencia entre ambas condiciones son similares

```python
import pandas as pd
from scipy import stats

# --- H3: Kruskal-Wallis — efecto de condición (Reposo / Izquierda / Derecha) ---
print("=" * 75)
print("H3 — KRUSKAL-WALLIS: EFECTO DE CONDICIÓN DENTRO DE CADA TAREA")
print("Prueba: Kruskal-Wallis (3 grupos)  |  α = 0.05")
print("=" * 75)

filas3 = []

for tipo in ['Motion', 'Imagery']:
    sub = df_eeg[df_eeg['Tarea'] == tipo]
    
    for ch in ['C3', 'C4']:
        for banda in ['Mu', 'Beta']:

            groups = [
                sub[(sub['Condicion'] == c) & (sub['Canal'] == ch)][banda].dropna().values
                for c in ['Reposo', 'Izquierda', 'Derecha']
            ]
            

            if any(len(g) < 5 for g in groups):
                continue
            

            stat, p = stats.kruskal(*groups)
            dec = "✓ Rechaza H0" if p < 0.05 else "— No rechaza H0"
            
            filas3.append({
                'Tipo': tipo, 
                'Canal': ch, 
                'Banda': banda,
                'H_stat': round(stat, 2), 
                'p-valor': round(p, 4),
                'Decisión (α=0.05)': dec
            })

# Mostramos la tabla de resultados finales
df_h3 = pd.DataFrame(filas3)
display(df_h3)
```

El resultado de la prueba de Kruskal-Wallis muestra que sí existen diferencias estadísticamente significativas entre las condiciones (reposo, mano izquierda y mano derecha) dentro de la tarea de imaginación motora, tanto en los canales C3 y C4 como en las bandas mu y beta. Esto se evidencia porque en todos los casos el valor p es prácticamente 0 (menor a 0.05), lo que lleva a rechazar la hipótesis nula (H0). En otras palabras, al menos una de las condiciones presenta una distribución de potencia diferente respecto a las otras. Sin embargo, esta prueba no indica entre qué pares específicos de condiciones están las diferencias
